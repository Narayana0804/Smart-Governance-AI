```python
import pandas as pd
from bertopic import BERTopic
from sentence_transformers import SentenceTransformer

# 1. Load Data
df = pd.read_csv('complaints_processed.csv')
docs = df['clean_text'].dropna().tolist()

# 2. Generate Embeddings on GPU
print('Generating embeddings...')
embedding_model = SentenceTransformer('all-MiniLM-L6-v2', device='cuda')
embeddings = embedding_model.encode(docs, show_progress_bar=True)

# 3. Train BERTopic
print('Training BERTopic...')
topic_model = BERTopic(
    min_topic_size=20,
    calculate_probabilities=False,
    verbose=True
)
topics, probs = topic_model.fit_transform(docs, embeddings)

# 4. Save Model
import os
os.makedirs('topic_model', exist_ok=True)
topic_model.save('topic_model', serialization='safetensors', save_ctfidf=True)

print('✅ Training complete! Zip and download the `topic_model` folder.')
```
